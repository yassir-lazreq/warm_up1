
<H3><span style="font-weight: bold; color: #1a8bcdff;">🧪 Challenge diagnostique — Gestion d'une classe</span></H3>

1. <span style="font-weight: bold; color: #e56922ff;">Problème</span> : </br>
Vous devez développer un petit programme Python permettant à un formateur de gérer les résultats de ses étudiants.</br></br>

Le programme doit permettre au formateur de saisir plusieurs étudiants. Pour chaque étudiant, il doit enregistrer son <span style="font-weight: bold;">nom</span>, son <span style="font-weight: bold;">âge</span> et sa <span style="font-weight: bold;">note</span>.</br></br>

Le programme doit ensuite :</br>
* Afficher la liste des étudiants avec leurs informations.
* Déterminer si chaque étudiant est <span style="font-weight: bold;">admis</span> ou <span style="font-weight: bold;">non admis</span> selon sa note.
* Calculer la <span style="font-weight: bold;">moyenne de la classe</span>.
* Afficher l'étudiant ayant obtenu la <span style="font-weight: bold;">meilleure note</span>.
* Afficher le nombre d'étudiants <span style="font-weight: bold;">admis</span> et <span style="font-weight: bold;">non admis</span>.
* Sauvegarder les résultats dans un fichier <span style="font-weight: bold;">resultats.txt</span>.</br></br>

Le programme doit également vérifier les saisies de l'utilisateur. Une note doit être comprise entre <span style="font-weight: bold;">0 et 20</span> et une saisie incorrecte ne doit pas provoquer l'arrêt du programme.</br></br>

L'application doit fonctionner dans le terminal avec un petit menu permettant de choisir entre :</br>
1. Ajouter des étudiants
2. Afficher les étudiants
3. Afficher les statistiques
4. Sauvegarder les résultats
5. Quitter</br></br>

2. <span style="font-weight: bold; color: #e56922ff;">Contraintes</span> :</br>
* Utiliser uniquement Python et ses fonctionnalités natives.
* Organiser le programme avec plusieurs fonctions.
* Les étudiants doivent être stockés dans une <span style="font-weight: bold;">liste de dictionnaires</span>.
* Les données doivent être sauvegardées dans <span style="font-weight: bold;">resultats.txt</span>.
* Le programme doit gérer les erreurs simples de saisie.
