2. <span style="font-weight: bold; color: #e56922ff;">Travail à faire</span>:</br>
Réécris le programme du Challenge 2 dans une fonction appelée horaire_sup.

```python
#  Écrivez du code Python ici

def horaire_sup(name, salaire_par_heure , heurs):
    
    try :
        
        salaire_par_heure = float(salaire_par_heure)
        heurs = float(heurs)
        
        salaire_par_heure = abs(salaire_par_heure)
        heurs = abs(heurs)
        
        salaire_totale = 0
        if heurs > 40:
            salaire_totale = salaire_par_heure*40 + (heurs - 40) * salaire_par_heure * 1.5
        else:
            salaire_totale = salaire_par_heure * heurs
        print(f"l'employe {name} get {salaire_totale} $")
    except ValueError:
        print("enter valid imput")
    

horaire_sup("yassir", 300 , 50)
```
