1. <span style="font-weight: bold; color: #e56922ff;">Problème</span> : </br>
Vous devez développer un petit programme Python permettant à un formateur de gérer les résultats de ses étudiants.</br></br>

Le programme doit permettre au formateur de saisir plusieurs étudiants. Pour chaque étudiant, il doit enregistrer son <span style="font-weight: bold;">nom</span>, son <span style="font-weight: bold;">âge</span> et sa <span style="font-weight: bold;">note</span>.</br></br>

Le programme doit ensuite :</br>
* Afficher la liste des étudiants avec leurs informations.
* Déterminer si chaque étudiant est <span style="font-weight: bold;">admis</span> ou <span style="font-weight: bold;">non admis</span> selon sa note.
* Calculer la <span style="font-weight: bold;">moyenne de la classe</span>.
* Afficher l'étudiant ayant obtenu la <span style="font-weight: bold;">meilleure note</span>.
* Afficher le nombre d'étudiants <span style="font-weight: bold;">admis</span> et <span style="font-weight: bold;">non admis</span>.
* Sauvegarder les résultats dans un fichier <span style="font-weight: bold;">resultats.txt</span>.</br></br>

Le programme doit également vérifier les saisies de l'utilisateur. Une note doit être comprise entre <span style="font-weight: bold;">0 et 20</span> et une saisie incorrecte ne doit pas provoquer l'arrêt du programme.</br></br>

L'application doit fonctionner dans le terminal avec un petit menu permettant de choisir entre :</br>
1. Ajouter des étudiants
2. Afficher les étudiants
3. Afficher les statistiques
4. Sauvegarder les résultats
5. Quitter</br></br>

2. <span style="font-weight: bold; color: #e56922ff;">Contraintes</span> :</br>
* Utiliser uniquement Python et ses fonctionnalités natives.
* Organiser le programme avec plusieurs fonctions.
* Les étudiants doivent être stockés dans une <span style="font-weight: bold;">liste de dictionnaires</span>.
* Les données doivent être sauvegardées dans <span style="font-weight: bold;">resultats.txt</span>.
* Le programme doit gérer les erreurs simples de saisie.



```python
#  Écrivez du code Python ici

def main():
    students = []
    while True:
        print("veuillez choisir une option: \n")
        print("1. ajouter un etudiant\n")
        print("2. afficher la liste des etudiants\n")
        print("3. afficher les statistiques\n")
        print("4. sauvegarder les resultats\n")
        print("5. quitter\n\n")
        
        choice = input("Entrez votre choix: ")
        
        match choice:
            case "1":
                add_student(students)
            case "2":
                display_students(students)
            case "3":
                display_statistics(students)
            case "4":
                save_results(students)
            case "5":
                print("Au revoir!")
                break
            case _:
                print("Choix invalide. Veuillez réessayer.\n")

def add_student(students):
    try:
        name = input("Entrez le nom de l'étudiant: ")
        
        if not name:
            print("Erreur: le nom ne peut pas être vide.\n")
            return True
        
        age = int(input("Entrez l'âge de l'étudiant: "))
        
        if age < 0:
            print("Erreur: l'âge ne peut pas être négatif.\n")
            return True
        
        grade = float(input("Entrez la note de l'étudiant: "))
        
        if grade < 0 or grade > 20:
            print("Erreur: la note doit être entre 0 et 20.\n")
            return True

        student = {"name": name, "age": age, "grade": grade}
        students.append(student)
        print(f"Étudiant {name} ajouté avec succès!\n")
    except ValueError:
        print("Erreur: veuillez entrer des valeurs valides.\n")
    return True

def display_students(students):
    try:
        if not students:
            print("Aucun étudiant à afficher.\n")
            return True
        print("Liste des étudiants:\n\n")
        for student in students:
            print(f"Nom: {student['name']}, Âge: {student['age']}, Note: {student['grade']}")
            print("\n")
        return True
    except Exception as e:
        print(f"Erreur lors de l'affichage des étudiants: {e}\n")
        return False

def display_statistics(students):
    if not students:
        print("Aucun étudiant pour calculer les statistiques.\n")
        return True
    total_students = len(students)
    average_grade = sum(student['grade'] for student in students) / total_students
    first_student = max(students, key=lambda x: x['grade'])
    last_student = min(students, key=lambda x: x['grade'])
    print(f"Nombre total d'étudiants: {total_students}")
    print(f"Note moyenne: {average_grade:.2f}")
    print(f"Meilleur étudiant: {first_student['name']} ({first_student['grade']})")
    print(f"Pire étudiant: {last_student['name']} ({last_student['grade']})")
    return True

def save_results(students):
    try:
        with open("results.txt", "w") as file:
            for student in students:
                file.write(f"Nom: {student['name']}, Âge: {student['age']}, Note: {student['grade']}\n")
        print("Résultats sauvegardés avec succès dans 'results.txt'.\n")
        return True
    except Exception as e:
        print(f"Erreur lors de la sauvegarde des résultats: {e}\n")
        return False
```
